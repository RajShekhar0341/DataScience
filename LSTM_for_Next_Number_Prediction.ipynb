#1. Implement an LSTM network to predict the next number in a sequence.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
#TrainingData
X=np.array([[[1],[2],[3]],[[2],[3],[4]],[[3],[4],[5]],[[4],[5],[6]],[[5],[6],[7]]], dtype=float)
y = np.array([4, 5, 6, 7, 8], dtype=float)
```

```python
# LSTM model
model = Sequential([
    LSTM(10, input_shape=(3, 1)),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)

model.fit(X, y, epochs=500, verbose=0)

# Test
test = np.array([[[6], [7], [8]]], dtype=float)

prediction = model.predict(test)

print("Predicted next number:", prediction)
```

#3rd Example

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Training data
X = np.array([
    [[1], [2], [3]],
    [[2], [3], [4]],
    [[3], [4], [5]],
    [[4], [5], [6]],
    [[5], [6], [7]],
    [[6], [7], [8]]
], dtype=np.float32)

y = np.array([4, 5, 6, 7, 8, 9], dtype=np.float32)

# Normalize
X_scaled = X / 10.0
y_scaled = y / 10.0

# Model
model = Sequential([
    LSTM(50, input_shape=(3, 1)),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.fit(
    X_scaled,
    y_scaled,
    epochs=3000,
    verbose=0
)

# Test
test = np.array([[[6], [7], [8]]], dtype=np.float32)
test_scaled = test / 10.0

prediction = model.predict(test_scaled, verbose=0)

# Convert back
predicted_value = prediction[0][0] * 10

print("Raw LSTM prediction:", predicted_value)
print("Predicted next number:", round(float(predicted_value)))
```

#2. LSTM for Increasing/Decreasing Sequence Classification

Implement an LSTM to classify whether a sequence is increasing or decreasing.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python

X = np.array([
    [[1], [2], [3]],
    [[2], [3], [4]],
    [[3], [4], [5]],
    [[3], [2], [1]],
    [[4], [3], [2]],
    [[5], [4], [3]]
], dtype=float)

```

```python
y = np.array([1,1,1,0,0,0])
model = Sequential([
    LSTM(8, input_shape=(3,1)),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=200, verbose=0)

# Test sequence
test = np.array([[[6], [7], [8]]], dtype=float)

prediction = model.predict(test)

if prediction[0][0] >= 0.5:
  print("Increasing Sequence")
else:
    print("Decreasing Sequence")
```

#3. LSTM for Simple Sentiment Analysis

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences = [
    "I love this movie",
    "This movie is excellent",
    "I like this film",
    "This is very good",
    "I hate this movie",
    "This movie is terrible",
    "I dislike this film",
    "This is very bad"
]
```

```python
# 1 = Positive
# 0 = Negative
labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])

# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)

X = pad_sequences(sequences, maxlen=5)

vocab_size = len(tokenizer.word_index) + 1

```

```python
# LSTM model
model = Sequential([
    Embedding(vocab_size, 8),
    LSTM(16),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X, labels, epochs=100, verbose=0)
```

```python
# Test
test_sentence = ["I love this film"]

test_seq = tokenizer.texts_to_sequences(test_sentence)
test_seq = pad_sequences(test_seq, maxlen=5)

prediction = model.predict(test_seq)

if prediction[0][0] >= 0.5:
    print("Positive")
else:
    print("Negative")
```

#4. LSTM for Spam Message Classification

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
messages = [
    "win free prize now",
    "claim free gift today",
    "you won cash prize",
    "free offer available",
    "meeting at ten tomorrow",
    "please submit your report",
    "class starts at nine",
    "call me after lunch"
]
```

```python
labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])

tokenizer = Tokenizer()
tokenizer.fit_on_texts(messages)

sequences = tokenizer.texts_to_sequences(messages)

X = pad_sequences(sequences, maxlen=5)

vocab_size = len(tokenizer.word_index) + 1

model = Sequential([
    Embedding(vocab_size, 8),
    LSTM(16),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X, labels, epochs=100, verbose=0)

# Test
test_message = ["claim free prize"]

test_seq = tokenizer.texts_to_sequences(test_message)
test_seq = pad_sequences(test_seq, maxlen=5)

prediction = model.predict(test_seq)

if prediction[0][0] >= 0.5:
    print("Spam")
else:
    print("Not Spam")
```
