LangChain program to create prompt template to generate a short explanation of user-supplied topic

```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template(
    "Explain {topic} in 4 simple points for engineering students."
)

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

chain = prompt | model | StrOutputParser()

topic = input("Enter a topic: ")

result = chain.invoke({"topic": topic})

print(result)
```

Creating a LangChain application that gives differnce between Machine Learning and Deep Learning in two-column format. *italicized text*

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from IPython.display import display, HTML

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

prompt = """
Compare Machine Learning and Deep Learning.

Create a neat HTML table with exactly two columns:
Machine Learning and Deep Learning.

Compare them based on:
Definition, Data Requirement, Feature Engineering,
Computational Requirement, Training Time,
Interpretability, Suitable Problems and Use Cases.

In each cell, show the comparison point in bold,
followed by a short explanation.
Give 3 practical use cases for each.

Return only the HTML table.
Do not use Markdown or code fences.
"""

response = llm.invoke(prompt)

display(HTML(response.text))
```

LCEL chain containing a Prompt Template, a Chat Model, and an output parser.

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

prompt = PromptTemplate.from_template(
    "Explain {topic} in 3 simple points for engineering students."
)

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

parser = StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({"topic": "Deep Learning"})

print(result)
```

LLM-Based Teaching Assistant Using Different Prompt Types

```python
prompts = {
    "concept": """
    You are an experienced Machine Learning faculty member.
    Explain overfitting to engineering students using a simple example.
    Keep the explanation within 5 points.
    """,

    "quiz": """
    Generate 5 MCQs on Machine Learning.
    Give four options and the correct answer for each question.
    """,

    "assignment": """
    Create a Machine Learning assignment on classification.
    Give 3 application-oriented questions of moderate difficulty.
    """
}

task = input("Enter task (concept / quiz / assignment / doubt): ")

if task == "doubt":
    doubt = input("Enter your Machine Learning doubt: ")

    prompt = f"""
    You are a Machine Learning teaching assistant.

    Student doubt:
    {doubt}

    Explain it in simple language with one suitable example.
    """
else:
    prompt = prompts[task]

response = llm.invoke(prompt)

print(response.text)
```

AI-Powered Question Paper Generation using LangChain and LCEL

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

prompt = PromptTemplate.from_template("""
Generate questions for the subject {subject}.

Syllabus/Topics:
{syllabus}

Difficulty Level: {difficulty}
Question Type: {question_type}
Number of Questions: {number}

Generate exactly the specified number of questions.
Restrict the questions to the given syllabus.
Match the specified difficulty and question type.
Avoid duplicate questions.
""")

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

chain = prompt | model | StrOutputParser()

subject = input("Enter Subject: ")
syllabus = input("Enter Syllabus/Topics: ")
difficulty = input("Enter Difficulty Level: ")
question_type = input("Enter Question Type: ")
number = input("Enter Number of Questions: ")

result = chain.invoke({
    "subject": subject,
    "syllabus": syllabus,
    "difficulty": difficulty,
    "question_type": question_type,
    "number": number
})

print(result)
```

Reusable Prompt Template for Java Program Generation

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

prompt = PromptTemplate.from_template("""
Write a Java program for the following problem:

{problem}

Requirements:
- Use simple Java syntax.
- Include necessary input and output statements.
- Give only the program with a short explanation.
""")

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

chain = prompt | model | StrOutputParser()

problem = input("Enter the Java problem statement: ")

result = chain.invoke({
    "problem": problem
})

print(result)
```

Professional Email Generator using One-Shot Prompting

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template("""
Convert the user's informal message into a professional email.

Example:

Informal Message:
I cannot attend class tomorrow because I am sick.

Professional Email:
Subject: Request for Leave

Dear Sir/Madam,
I am unable to attend class tomorrow due to illness.
Kindly grant me leave for the day.

Regards,
Student

Now convert the following message:

Informal Message:
{message}

Professional Email:
""")

chain = prompt | model | StrOutputParser()

message = input("Enter an informal message: ")

result = chain.invoke({"message": message})

print(result)
```

Sentiment Analyzer using Few-Shot Prompting

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template("""
Classify the sentiment as Positive, Negative, or Neutral.

Examples:

Text: The course was excellent and very easy to understand.
Sentiment: Positive

Text: The application crashes frequently and is frustrating.
Sentiment: Negative

Text: The class was conducted for two hours.
Sentiment: Neutral

Now classify:

Text: {text}

Give the sentiment and a one-line reason.
""")

chain = prompt | model | StrOutputParser()

text = input("Enter a review or statement: ")

result = chain.invoke({"text": text})

print(result)
```
